```python
import numpy as np
from scipy import stats
```

```python
# 1. Membuat data contoh (100 sampel acak, misalnya dari distribusi normal dengan mean=50 dan std=10)
# Anda bisa mengganti bagian ini dengan data Anda sendiri
np.random.seed(42)
data = np.random.normal(loc=50, scale=10, size=100)

print(f"Jumlah sampel: {len(data)}")
print(f"Data awal (5 sampel pertama): {data[:5]}\n")
```

```python
# BAGIAN A: Estimasi Titik dan Interval
# Estimasi titik: Rata-rata sampel (mean)
titik_estimasi = np.mean(data)
# Tingkat kesalahan (alpha) = 5% -> Tingkat kepercayaan = 95% (0.95)
alpha = 0.05
n = len(data)
std_dev = np.std(data, ddof=1) # Standar deviasi sampel
# Menggunakan distribusi Z (karena n = 100 cukup besar, mendekati distribusi normal)
z_score = stats.norm.ppf(1 - alpha/2)
margin_of_error = z_score * (std_dev / np.sqrt(n))

interval_bawah = titik_estimasi - margin_of_error
interval_atas = titik_estimasi + margin_of_error

print("--- A. Estimasi Titik dan Interval (Z-Distribution, Kesalahan 5%) ---")
print(f"Nilai Estimasi Titik (Mean): {titik_estimasi:.4f}")
print(f"Interval Kepercayaan 95%: [{interval_bawah:.4f}, {interval_atas:.4f}]\n")
```

```python
# BAGIAN B: Resampling Jackknife & Bootstrapping
# 1. Metode Jackknife
# Mengeluarkan 1 data secara bergantian sebanyak n kali
jackknife_means = []
for i in range(n):
  # Membuat sampel baru dengan membuang data ke-i
    sample_jack = np.delete(data, i)
    jackknife_means.append(np.mean(sample_jack))

rata_rata_jackknife = np.mean(jackknife_means)
print("--- B1. Metode Jackknife ---")
print(f"Nilai rata-rata dari hasil resampling Jackknife: {rata_rata_jackknife:.4f}\n")
```

```python
# 2. Metode Bootstrapping (1000 iterasi)
n_iterations = 1000
bootstrap_means = []

for i in range(n_iterations):
    # Mengambil sampel dengan pengembalian (resampling) sebanyak ukuran data asli
    sample_boot = np.random.choice(data, size=n, replace=True)
    bootstrap_means.append(np.mean(sample_boot))

rata_rata_bootstrap = np.mean(bootstrap_means)
print("--- B2. Metode Bootstrapping ---")
print(f"Jumlah Iterasi: {n_iterations}")
print(f"Nilai rata-rata dari hasil resampling Bootstrapping: {rata_rata_bootstrap:.4f}")
```
